# 🎓 مشروع ELM - نظام إدارة الطلاب الذكي

## 📌 ملخص المشروع الشامل

### الفكرة العامة
نظام برمجيات ذكي وشامل مخصص للمدارس والمعاهد الأهلية يدعم:
- إدارة **1000+ طالب و100+ مدرس** لكل مؤسسة
- جميع الأنظمة التعليمية (بولونيا، كورسات، جامعة أمريكية، مدارس عادية)
- نموذجي تشغيل: **SaaS سحابي** أو **On-Premise محلي**
- أتمتة كاملة للعمليات الإدارية والمالية والأكاديمية

## 1️⃣ البرومبت الشامل للمبرمج/الذكاء الاصطناعي

### استخدم هذا البرومبت الكامل عند إعطاء المشروع لأي AI أو مبرمج:

```
أنت الآن مبرمج Full-Stack خبير ومهندس معمارية برمجيات (Software Architect). 
أطلب منك بناء منصة متكاملة (SaaS Multi-Tenant Educational Management System) 
لشركة برمجيات اسمها (ELM).

━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

🎯 الفكرة العامة للمشروع (Project Vision & Overview)
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

هذا المشروع عبارة عن نظام برمجيات ذكي وشامل مخصص للمدارس والمعاهد الأهلية 
التي قد تضم المؤسسة الواحدة منها أكثر من 1000 طالب و100 مدرس.

النظام مستوحى من الآليات الحديثة لإدارة التعليم، مثل "نظام مسار بولونيا"، 
ويهدف إلى رقمنة وأتمتة كافة العمليات الإدارية، المالية، والأكاديمية، 
مما يوفر على الإدارات مئات الساعات من العمل اليدوي.

النظام مصمم ليكون "صندوقاً سحرياً متكاملاً" يحل مشاكل الإدارات، 
ويتميز بتوفير:
- لوحات تحكم متطورة للإدارة والأساتذة
- تطبيق هاتف ذكي للطلاب وأولياء الأمور
- حلقة وصل متكاملة تدير المؤسسة بكفاءة عالية

كما أنه يعتمد على نموذجين مرنين للتشغيل:
1️⃣ خدمة سحابية باشتراك شهري (SaaS) - $200/شهر
2️⃣ نظام محلي دائم (On-Premise) - $1000 مرة واحدة + دعم سنوي

━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

🔧 التقنيات ومرونة قاعدة البيانات (Tech Stack & Database Flexibility)
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

الخلفية البرمجية (Backend):
- اللغة: Python 3.9+
- الإطار: Flask
- Architecture: RESTful APIs

التعامل مع قاعدة البيانات (ORM):
- استخدم مكتبة SQLAlchemy حصراً لتكون قاعدة البيانات متغيرة (Dynamic)

بيئة التطوير (Development):
- يجب أن يعتمد النظام مبدئياً على SQLite لسهولة التطوير والتجريب

بيئة الإنتاج (Production):
- يجب أن يعتمد النظام على MySQL مع تفعيل محرك InnoDB
- ضمان خاصية (WAL) وحماية البيانات من التلف عند الانقطاع المفاجئ

آلية التحويل بين البيئات:
- يجب أن يتم التحويل بين SQLite و MySQL فقط من خلال تغيير رابط قاعدة البيانات
- المتغير: SQLALCHEMY_DATABASE_URI داخل ملف .env
- دون تغيير سطر كود واحد في النماذج (Models)

الواجهات (Frontend):
- لوحة تحكم الإدارة: HTML/CSS/JavaScript (عصرية ومتجاوبة)
- تطبيق الموبايل: Flutter (يعمل على Android و iOS من كود واحد)

الأمن والحماية:
- المصادقة: JWT (JSON Web Tokens) + Secure Cookies
- تعمية الأكواد: PyArmor (حماية حقوق الملكية)
- CORS و HTTPS

━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

📊 محرك الأنظمة التعليمية الشامل (Dynamic Educational Engine)
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

يجب تصميم جداول قاعدة البيانات لتكون مرنة (Dynamic Settings).

عند إنشاء مدرسة جديدة، يختار الإدمن نوع النظام التعليمي وتتغير الواجهات:

1. نظام مسار بولونيا والجامعة الأمريكية (Credit Hours)
   - إدارة الساعات المعتمدة
   - أوزان المواد
   - النتائج بناءً على الساعات

2. النظام الفصلي ونظام الكورسات
   - فصول دراسية (Semesters)
   - كورسات منفصلة (Courses)
   - مكونات التقييم (Assessment Components)

3. نظام المدارس والمعاهد العادية الكلاسيكية
   - فصول دراسية (Classes)
   - حصص أسبوعية ثابتة
   - درجات بسيطة

4. نظام الجدولة (Timetable) - ذكي ومتقدم
   - يربط: (المادة + الأستاذ + القاعة + وقت الدرس بالدقيقة)
   - منع التضارب الزمني تلقائياً
   - جدول أسبوعي وفصلي

━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

🔐 صفحة تسجيل الدخول والجلسات (Smart Login & Sessions)
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

1. تحديد نوع المستخدم (Role-Based Access Control)
   - خيار 1: هل أنت طالب؟
   - خيار 2: هل أنت تدريسي؟
   - هذا يسهل توجيه المستخدم إلى لوحة التحكم الصحيحة فوراً

2. حقل البحث عن المدرسة (Autocomplete Searchable Dropdown)
   - حقل بحث ذكي يظهر قائمة منسدلة
   - بمجرد كتابة أول حرفين، يبحث في قائمة المدارس
   - يظهر الاسم والرمز (Code) للاختيار السريع
   - تقليل أخطاء الكتابة (تصحيح تلقائي)

3. نظام المصادقة الآمن
   - استخدام JWT (JSON Web Tokens) للتوكن
   - تخزين التوكن في Secure Cookies
   - الجلسات طويلة الأمد (Long-lived Sessions)
   - تجديد التوكن تلقائياً قبل انتهاؤه

━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

🎛️ إدارة لوحة تحكم شركة ELM (Super Admin Dashboard)
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

هذه اللوحة مخصصة لك كمبرمج ومالك شركة ELM:

1. إنشاء المؤسسات (Tenants Management)
   - إضافة مدرسة أو معهد جديد
   - تعيين اسم المؤسسة
   - رمز المؤسسة (Organization Code)

2. الهوية البصرية (Branding)
   - الألوان الأساسية والثانوية
   - الشعار (Logo)
   - الخط المستخدم
   - هذا ينعكس على جميع الواجهات والتطبيقات الخاصة بالمؤسسة

3. تحديد نوع النظام التعليمي
   - اختيار من القائمة: بولونيا، كورسات، مدارس عادية، إلخ
   - تغيير نوع النظام لاحقاً ممكن لكن يحتاج تغيير البيانات

4. إدارة الاشتراكات
   - متابعة أي المؤسسات مدفوعة
   - أي المؤسسات متأخرة
   - إمكانية إيقاف النظام عن بعد للمتأخرين
   - إصدار الفواتير تلقائياً

5. توليد حسابات مديري المعاهد
   - بعد إنشاء المؤسسة، توليد حساب Super Admin للمؤسسة
   - إرسال بيانات الدخول للعميل (الإيميل والرقم السري المؤقت)
   - تفعيل حساب العميل تلقائياً

6. نقل المؤسسات من SaaS إلى On-Premise
   - عند طلب العميل، تحويل المؤسسة من السحابة إلى محلي
   - توليد Docker Container خاص بالمؤسسة
   - نقل بيانات المؤسسة تلقائياً

━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

🖥️ هندسة السيرفر المحلي والتحكم عن بعد (On-Premise Infrastructure)
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

في حال طلب العميل السيرفر المحلي:

1. الآلية التلقائية
   - العميل يختار "On-Premise" من لوحة Super Admin
   - النظام ينشئ Docker Compose خاص به
   - نقل البيانات الكاملة تلقائياً
   - توليد نطاق فرعي (Subdomain)

2. العتاد المطلوب
   - Raspberry Pi 5 (8GB RAM)
   - SSD خارجي 1TB (M.2 NVMe)
   - مزود طاقة أصلي (27W)
   - مروحة تبريد معدنية (Active Cooler)
   - جهاز Mini UPS (بطارية احتياطية 6-10 ساعات)

3. نظام التشغيل
   - Linux بنسخة Read-Only Overlay
   - يمنع تلف ملفات الإقلاع عند انقطاع الكهرباء
   - جميع البيانات والتطبيقات على الهارد الخارجي (Read-Write)

4. التخزين والنسخ الاحتياطي
   - مسارات قاعدة البيانات: /mnt/external/database/
   - مسارات الصور والملفات: /mnt/external/media/
   - سكريبت يومي (Cron Job) لضغط النسخة الاحتياطية
   - رفع النسخة السحابية إلى AWS S3 أو Google Cloud

5. التحكم عن بعد
   - استخدام Cloudflare Tunnels للربط الآمن
   - أو استخدام Tailscale لـ VPN آمن
   - صيانة بدون الحاجة لزيارة موقع المؤسسة
   - رفع التحديثات والتصحيحات الأمنية بدون تأثير على العملاء

6. النطاقات الفرعية (Subdomains)
   - الدومين الأساسي: backend-elm.com
   - نطاق فرعي لكل مؤسسة: schoolname.backend-elm.com
   - هذا يوفر عزلاً تاماً لملفات Cookies والجلسات
   - منع تداخل البيانات بين المؤسسات

7. حماية الكود
   - قبل التسليم، تعمية جميع أكواد Python باستخدام PyArmor
   - حماية الملكية الفكرية
   - منع النسخ غير القانوني

━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

✅ الخلاصة:

بناءً على هذا البرومبت الشامل، يجب أن ينتج عنه:

1. ✅ نماذج قاعدة البيانات الكاملة (SQLAlchemy Models)
2. ✅ جميع APIs المطلوبة (Flask Blueprints)
3. ✅ نظام المصادقة والتفويض (JWT + Role-Based Access)
4. ✅ لوحات التحكم (Web Interface)
5. ✅ تطبيق الموبايل (Flutter App)
6. ✅ Docker Compose للنشر
7. ✅ الحماية والتشفير الكامل
8. ✅ دعم Multi-Tenant Architecture
9. ✅ مرونة التحويل بين SQLite و MySQL
10. ✅ التحكم عن بعد والنسخ الاحتياطية

ابدأ بإنشاء هيكلية المشروع والنماذج الأساسية أولاً.
"""
```
